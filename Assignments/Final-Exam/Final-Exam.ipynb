4. Why might a full factorial design be preferred in this specific instance to a $2^k$ fractional factorial design?

- In particular, designs with exactly two levels for each factor are denoted by the $2^k$ notation. The "Protein" element in this experiment has three levels: bacon, sausage, and chorizo. A 3-level factor is theoretically incompatible with a normal $2^k$ fractional design. Although there are mixed-level designs, a full factorial is the most logical option due to the 3x2x2 structure's simplicity. The relationship between components is frequently the most crucial element in culinary science. A Hot salsa, for instance, can go well with Bacon but dominate the taste of Sausage. When there are so many elements that it is either monetarily or physically impracticable to test every combination, fractional factorial designs are usually employed (normally, 10 factors would require 1,024 runs). We want the most precise information because the objective is to identify the best ingredients and minimize waste by creating the ideal menu. To determine which particular burrito layout is best, a full factorial design offers the most data and statistical power.

7. Fit the full factorial ANOVA model. Check the assumption of normality using a testing method and a visual method. Does the assumption hold? Why or why not?

```python
## Question 7 Code ##

import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
import scipy.stats as stats
import matplotlib.pyplot as plt

# Fit the full factorial ANOVA model
# Formula: Satisfaction ~ Protein * Salsa * Potatoes (this includes main effects and all interactions)
bust_model = ols('Satisfaction ~ C(Protein) * C(Salsa) * C(Potatoes)', data=bust).fit()
anova_table = sm.stats.anova_lm(bust_model, typ=2)

# Extract residuals
residuals = bust_model.resid

# Shapiro-Wilk Test
shapiro_test = stats.shapiro(residuals)

# Q-Q Plot
plt.figure(figsize=(8, 6))
stats.probplot(residuals, dist="norm", plot=plt)
plt.title('Q-Q Plot of Residuals')
plt.show()
plt.close()

# Histogram of Residuals
plt.figure(figsize=(8, 6))
plt.hist(residuals, bins=15, edgecolor='black', alpha=0.7)
plt.title('Histogram of Residuals')
plt.xlabel('Residual Value')
plt.ylabel('Frequency')
plt.show()
plt.close()

# Output the results for analysis
print("--- ANOVA Table ---")
print(anova_table)
print("\n--- Shapiro-Wilk Test for Normality ---")
print(f"Statistic: {shapiro_test.statistic:.4f}, p-value: {shapiro_test.pvalue:.4f}")
```

The assumption for normality holds **true**

- In the Q-Q plot, the data points stay relatively close to the reference line, and the histogram appears approximately symmetric and unimodal. While there is minor noise in the tails, it is well within the acceptable range for a sample size of $n=60$. 8. Check the assumption of constant variance using a testing method and a visual method. Does the assumption hold? Why or why not?

```python
## Question 8 Code ##

import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from statsmodels.stats.diagnostic import het_breuschpagan

# Refit the model
model = ols('Satisfaction ~ C(Protein) * C(Salsa) * C(Potatoes)', data=bust).fit()

# Get residuals and fitted values
residuals = model.resid
fitted_values = model.fittedvalues

# Residuals vs. Fitted Plot
plt.figure(figsize=(10, 6))
plt.scatter(fitted_values, residuals, alpha=0.7)
plt.axhline(0, color='red', linestyle='--')
plt.title('Residuals vs. Fitted Values')
plt.xlabel('Fitted Values (Predicted Satisfaction)')
plt.ylabel('Residuals')
plt.show()
plt.close()

# Boxplot of Residuals by Group
bust['Group'] = bust['Protein'] + "_" + bust['Salsa'] + "_" + bust['Potatoes']
plt.figure(figsize=(12, 6))
sns.boxplot(x='Group', y=residuals, data=bust)
plt.xticks(rotation=45)
plt.title('Residuals by Treatment Combination')
plt.show()
plt.close()

# Levene's Test
groups = [group['Satisfaction'].values for name, group in bust.groupby(['Protein', 'Salsa', 'Potatoes'])]
levene_stat, levene_p = stats.levene(*groups)

print(f"Levene's Test: Statistic={levene_stat:.4f}, p-value={levene_p:.4f}")
```

The assumption for constant variance holds **true**

- While the boxplots show some variation in the "stretch" of the data for specific burrito combinations, the Residuals vs. Fitted plot does not show a strong "fan" or "funnel" shape that would indicate a systematic violation of homoscedasticity. The p-value for Levene's test ($0.2322$) is well above the common significance threshold of $0.05$. This means we fail to reject the null hypothesis of equal variances across groups.

11. Calculate and interpret partial $\eta^2$ for the main effects.

```python
## Question 11 Code ##

import pandas as pd
import statsmodels.api as sm
from statsmodels.formula.api import ols

# ANOVA model
bust_model = ols('Satisfaction ~ C(Protein) * C(Salsa) * C(Potatoes)', data=bust).fit()
anova_table = sm.stats.anova_lm(bust_model, typ=2)

# Define a function to calculate Partial Eta Squared
def calculate_partial_eta_squared(anova_df):
    ss_error = anova_df.loc['Residual', 'sum_sq']
    # Calculate for each effect in the table (excluding Residual)
    anova_df['partial_eta_sq'] = anova_df['sum_sq'] / (anova_df['sum_sq'] + ss_error)
    return anova_df

# Apply the calculation
results_with_eta = calculate_partial_eta_squared(anova_table)

# Display the main effects
main_effects_eta = results_with_eta.loc[['C(Protein)', 'C(Salsa)', 'C(Potatoes)'], ['sum_sq', 'partial_eta_sq']]
print(main_effects_eta)
```

- Partial eta squared ($\eta_p^2) = \frac{SS_{effect}}{SS_{effect} + SS_{error}}$:

- Partial eta squared ($\eta_p^2$) measures the proportion of total variation in the dependent variable (Satisfaction) that is uniquely explained by a specific independent variable, after accounting for other factors and interactions in the model.

- Approximately **14.3%** of the variance in customer satisfaction that is not explained by other factors can be attributed to the type of protein used. This confirms that the choice of Bacon, Sausage, or Chorizo is the most critical driver of burrito quality in this experiment.

- About **7.8%** for Salsa and **6%** for Potatoes; the effect size suggests it still plays a moderate role in how customers perceive the burrito, but t has a smaller practical impact on satisfaction than the choice of protein.
